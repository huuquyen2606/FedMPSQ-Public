# Run the reported FedMPSQ arm on Kaggle

The reported byte cap is selected automatically from `NUM_CLIENTS`: 3,334 bytes for 10 clients and 3,044 bytes for 100 clients. Planning is the default.

```python
from pathlib import Path
import subprocess
import sys

NUM_CLIENTS = 10  # 10 or 100
SEED = 42
EXECUTE = False
EVALUATE_TEST = True

REPO_ROOT = Path('/kaggle/working/ids_fed')
DATA_DIRS = {
    10: Path('/kaggle/input/your-dataset/prepared-10'),
    100: Path('/kaggle/input/your-dataset/prepared-100'),
}
DATA_DIR = DATA_DIRS[NUM_CLIENTS]
MANIFEST = REPO_ROOT / f'configs/partitions/{NUM_CLIENTS}_clients_manifest.json'
OUTPUT = Path('/kaggle/working/results/fedmpsq')
```

```python
ARMS = {10: 'rht_g4_b3334', 100: 'rht_g4_b3044_c100'}
if NUM_CLIENTS not in ARMS:
    raise ValueError('NUM_CLIENTS must be 10 or 100')
for path in (REPO_ROOT, DATA_DIR, MANIFEST):
    if not path.exists():
        raise FileNotFoundError(path)
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', '-e', str(REPO_ROOT)], check=True)
```

```python
command = [
    sys.executable, str(REPO_ROOT / 'scripts/run_uplink_v4.py'),
    '--data-dir', str(DATA_DIR),
    '--manifest', str(MANIFEST),
    '--output', str(OUTPUT),
    '--clients', str(NUM_CLIENTS),
    '--arm', ARMS[NUM_CLIENTS],
    '--seed', str(SEED),
    '--seeds', str(SEED),
]
if EXECUTE:
    command.append('--run')
    if EVALUATE_TEST:
        command.append('--evaluate-test')
print(' '.join(command))
subprocess.run(command, cwd=REPO_ROOT, check=True)
```
